## 1 · Peek at the data

Always look at your data with your own eyes before writing checks. What smells wrong?

```python
import duckdb, pandas as pd
con = duckdb.connect('../avocados.duckdb')
df = con.execute('SELECT * FROM orders').df()
print(df.shape)
df.head(10)
```

```python
# quick profile — pandas describe covers a lot for free
df.describe(include='all').T
```

## 2 · Run the starter Soda scan from Python

Same result as `soda scan` on the CLI — the framework is a Python package underneath.

```python
import os
os.environ['SODA_TELEMETRY_DISABLED'] = 'true'

from soda.scan import Scan

scan = Scan()
scan.set_data_source_name('avocados')
scan.add_configuration_yaml_file('../soda/configuration.yml')
scan.add_sodacl_yaml_file('../soda/checks_starter.yml')
scan.execute()

print(scan.get_logs_text())
```

## 3 · Now the extended checks

```python
scan = Scan()
scan.set_data_source_name('avocados')
scan.add_configuration_yaml_file('../soda/configuration.yml')
scan.add_sodacl_yaml_file('../soda/checks_extended.yml')
scan.execute()
print(scan.get_logs_text())
```

## 4 · Bonus — profile the table with `ydata-profiling`

Automated profiling gives you a full HTML report of columns, distributions, correlations, missing values. Great for **discovering** what checks you should write in the first place.

```python
from ydata_profiling import ProfileReport

profile = ProfileReport(df, title='Avocado orders — profile', minimal=True)
profile.to_notebook_iframe()
```
